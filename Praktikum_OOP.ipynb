```python
# Latihan 1 - Membuat class hero

class Hero:
    # Constructor
    def __init__(self, name, hp, attack_power):
        self.name = name
        self.hp = hp
        self.attack_power = attack_power

    # Method untuk menampilkan informasi hero
    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")


# Membuat object
hero1 = Hero("Layla", 100, 15)
hero2 = Hero("Zilong", 120, 20)

# Menampilkan informasi
hero1.info()
hero2.info()

# Analisis 1
hero1.hp = 500
print("HP Layla setelah diubah:", hero1.hp)
```

```python
# Latihan 2 - Interaksi Antar Object

class Hero:
    def __init__(self, name, hp, attack_power):
        self.name = name
        self.hp = hp
        self.attack_power = attack_power

    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

    # Method menyerang
    def serang(self, lawan):
        print(f"{self.name} menyerang {lawan.name}!")
        lawan.diserang(self.attack_power)

    # Method menerima serangan
    def diserang(self, damage):
        self.hp -= damage
        print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")


hero1 = Hero("Layla", 100, 15)
hero2 = Hero("Zilong", 120, 20)

print("--- Pertarungan Dimulai ---")

hero1.serang(hero2)
hero2.serang(hero1)
```

Maksud:

ANALISIS 2

Kesimpulan:

```python
# Latihan 3 -Inheritance/Pewarisan

class Hero:
    def __init__(self, name, hp, attack_power):
        self.name = name
        self.hp = hp
        self.attack_power = attack_power

    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

    def serang(self, lawan):
        print(f"{self.name} menyerang {lawan.name}!")
        lawan.diserang(self.attack_power)

    def diserang(self, damage):
        self.hp -= damage
        print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")


class Mage(Hero):
    def __init__(self, name, hp, attack_power, mana):
        super().__init__(name, hp, attack_power)
        self.mana = mana

    def info(self):
        print(f"{self.name} [Mage] | HP: {self.hp} | Mana: {self.mana}")

    def skill_fireball(self, lawan):
        if self.mana >= 20:
            print(f"{self.name} menggunakan Fireball ke {lawan.name}!")
            self.mana -= 20
            lawan.diserang(self.attack_power * 2)
        else:
            print(f"{self.name} gagal skill! Mana tidak cukup.")


print("--- Update Class Hero ---")

eudora = Mage("Eudora", 80, 30, 100)
balmond = Hero("Balmond", 200, 10)

eudora.info()
eudora.serang(balmond)
eudora.skill_fireball(balmond)
```

```python
#Latihan 4 - Encapsulation

class Hero:
    def __init__(self, nama, hp_awal):
        self.nama = nama
        self.__hp = hp_awal

    # Getter
    def get_hp(self):
        return self.__hp

    # Setter
    def set_hp(self, nilai_baru):
        if nilai_baru < 0:
            self.__hp = 0
        elif nilai_baru > 1000:
            print("Cheat terdeteksi! HP dimaksimalkan ke 1000 saja.")
            self.__hp = 1000
        else:
            self.__hp = nilai_baru

    def diserang(self, damage):
        sisa_hp = self.get_hp() - damage
        self.set_hp(sisa_hp)

        print(
            f"{self.nama} terkena damage {damage}. "
            f"Sisa HP: {self.get_hp()}"
        )


hero1 = Hero("Layla", 100)

hero1.set_hp(-50)

print("HP Layla:", hero1.get_hp())
```

Kesimpulan:
Encapsulation dan setter digunakan untuk melindungi serta
memvalidasi data agar nilainya tetap sesuai aturan.

```python
#Latihan 5 - Abstraction & Interface

from abc import ABC, abstractmethod


class GameUnit(ABC):

    @abstractmethod
    def serang(self, target):
        pass

    @abstractmethod
    def info(self):
        pass


class Hero(GameUnit):
    def __init__(self, nama):
        self.nama = nama

    def serang(self, target):
        print(f"Hero {self.nama} menebas {target}!")

    def info(self):
        print(f"Saya adalah Hero: {self.nama}")


class Monster(GameUnit):
    def __init__(self, jenis):
        self.jenis = jenis

    def serang(self, target):
        print(f"Monster {self.jenis} menggigit {target}!")

    def info(self):
        print(f"Saya adalah Monster: {self.jenis}")


h = Hero("Alucard")
m = Monster("Serigala")

h.info()
m.info()
```

ANALISIS 5

Jika method serang() pada Hero dihapus, program error karena Hero
belum memenuhi aturan dari abstract class GameUnit. GameUnit juga
tidak dapat dibuat menjadi object secara langsung.

Kesimpulan:
Abstraction digunakan sebagai blueprint atau aturan yang harus
dipenuhi oleh class turunannya.

```python
#Latihan 6 - Polymorphism

class Hero:
    def __init__(self, nama):
        self.nama = nama

    def serang(self):
        print("Hero menyerang dengan tangan kosong.")


class Mage(Hero):
    def serang(self):
        print(f"{self.nama} (Mage) menembakkan Bola Api! Boom!")


class Archer(Hero):
    def serang(self):
        print(f"{self.nama} (Archer) memanah dari jauh! Jleb!")


class Fighter(Hero):
    def serang(self):
        print(f"{self.nama} (Fighter) memukul dengan pedang! Slash!")


pasukan = [
    Mage("Eudora"),
    Archer("Miya"),
    Fighter("Zilong"),
    Mage("Gord")
]

print("--- PERANG DIMULAI ---")

for pahlawan in pasukan:
    pahlawan.serang()
```

ANALISIS 6

Healer dapat ditambahkan tanpa mengubah looping karena memiliki
method serang() yang sama. Jika serang() pada Archer diganti menjadi
tembak_panah(), program error karena looping masih memanggil serang().

Kesimpulan:
Polymorphism memungkinkan berbagai object menggunakan method yang
sama dengan perilaku yang berbeda.

```python

from abc import ABC, abstractmethod


# ==========================================
# ABSTRACT CLASS
# ==========================================

class KamarHotel(ABC):

    def __init__(self, nama_kamar, stok, harga_dasar):
        self.nama_kamar = nama_kamar
        self.__stok = stok
        self.__harga_dasar = harga_dasar

    # Getter stok
    def get_stok(self):
        return self.__stok

    # Getter harga dasar
    def get_harga_dasar(self):
        return self.__harga_dasar

    # Setter stok dengan validasi
    def tambah_stok(self, jumlah):
        if self.__stok + jumlah < 0:
            print(
                f"Gagal update stok {self.nama_kamar}! "
                f"Stok tidak boleh negatif ({jumlah})."
            )
        else:
            self.__stok += jumlah
            print(
                f"Berhasil menambahkan stok "
                f"{self.nama_kamar}: {self.__stok} unit."
            )

    @abstractmethod
    def tampilkan_detail(self):
        pass

    @abstractmethod
    def hitung_harga_total(self, jumlah_malam):
        pass


# ==========================================
# CLASS KAMAR DELUXE
# ==========================================

class KamarDeluxe(KamarHotel):

    def __init__(self, nama_kamar, stok, harga_dasar, fasilitas):
        super().__init__(nama_kamar, stok, harga_dasar)
        self.fasilitas = fasilitas

    def tampilkan_detail(self):
        print(
            f"[DELUXE] {self.nama_kamar} | "
            f"Fasilitas: {self.fasilitas}"
        )

    def hitung_harga_total(self, jumlah_malam):
        pajak = self.get_harga_dasar() * 10 / 100
        harga_per_malam = self.get_harga_dasar() + pajak
        return harga_per_malam * jumlah_malam


# ==========================================
# CLASS KAMAR STANDARD
# ==========================================

class KamarStandard(KamarHotel):

    def __init__(self, nama_kamar, stok, harga_dasar, kapasitas):
        super().__init__(nama_kamar, stok, harga_dasar)
        self.kapasitas = kapasitas

    def tampilkan_detail(self):
        print(
            f"[STANDARD] {self.nama_kamar} | "
            f"Kapasitas: {self.kapasitas}"
        )

    def hitung_harga_total(self, jumlah_malam):
        pajak = self.get_harga_dasar() * 5 / 100
        harga_per_malam = self.get_harga_dasar() + pajak
        return harga_per_malam * jumlah_malam


# ==========================================
# FUNGSI PROSES TRANSAKSI
# ==========================================

def proses_transaksi(daftar_pesanan):

    total_tagihan = 0

    print("\n--- STRUK PEMESANAN ---")

    for nomor, pesanan in enumerate(daftar_pesanan, start=1):

        kamar = pesanan["kamar"]
        jumlah_malam = pesanan["malam"]

        kamar.tampilkan_detail()

        harga_dasar = kamar.get_harga_dasar()

        if isinstance(kamar, KamarDeluxe):
            pajak = harga_dasar * 10 / 100
            jenis_pajak = "10%"
        else:
            pajak = harga_dasar * 5 / 100
            jenis_pajak = "5%"

        subtotal = kamar.hitung_harga_total(jumlah_malam)

        print(
            f"  Harga Dasar/Malam: Rp {harga_dasar:,.0f} | "
            f"Pajak({jenis_pajak}): Rp {pajak:,.0f}"
        )

        print(
            f"  Menginap: {jumlah_malam} malam | "
            f"Subtotal: Rp {subtotal:,.0f}"
        )

        total_tagihan += subtotal

    print("----------------------------------------")
    print(f"TOTAL TAGIHAN: Rp {total_tagihan:,.0f}")
    print("----------------------------------------")


# ==========================================
# MAIN PROGRAM
# ==========================================

print("--- SETUP DATA KAMAR ---")

deluxe = KamarDeluxe(
    "Kamar Deluxe Sea View",
    0,
    1_500_000,
    "Private Pool"
)

standard = KamarStandard(
    "Kamar Standard Superior",
    0,
    500_000,
    "2 Orang"
)

# Menambahkan stok Deluxe
deluxe.tambah_stok(10)

# Mencoba memasukkan stok negatif
standard.tambah_stok(-5)

# Menambahkan stok Standard
standard.tambah_stok(20)


# ==========================================
# PEMESANAN
# ==========================================

daftar_pesanan = [
    {
        "kamar": deluxe,
        "malam": 2
    },
    {
        "kamar": standard,
        "malam": 1
    }
]

proses_transaksi(daftar_pesanan)
```
